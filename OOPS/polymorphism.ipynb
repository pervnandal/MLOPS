### Method overriding

```python
# Parent class
class Animal():
    def speak(self):
        return "Animal Speaks"

# Derived Class 1
class Dog(Animal):
    def speak(self):
        return "Barks"

# Derived Class 2
class Cat(Animal):
    def speak(self):
        return "Meow"
```

```python
dog = Dog()
cat = Cat()
```

```python
dog.speak()
```

```python
cat.speak()
```

#### Inference method

```python
from abc import ABC, abstractmethod
```

```python
# Abstract Class
class Vehicle(ABC):
    @abstractmethod
    def engine(self):
        pass

# Derived Class 1
class Car(Vehicle):
    def engine(self):
            return "Car Engine Started"

# Derived Class 2
class Truck(Vehicle):
    def engine(self):
            return "Truck Engine Started"
```

```python
car = Car()
truck = Truck()
```

```python
car.engine()
```

```python
truck.engine()
```
