### Encapsulation

#### Variable types and syntax
- **public** - `self.variable_name`
- **private** - `self.__variable_name`
- **protected** - `self._variable_name`

```python
class Person():
    def __init__(self,name,age,gender):
        self._name = name # Protected - can be accessed within the parent and derived class
        self.__age = age # Private - can be accessed within the parent
        self.gender = gender # public - can be accessed from anywhere

class Employee(Person):
    def __init__(self,name,age,gender):
        super().__init__(name,age,gender)

emp = Employee("Parv",21,"Male")
```
